```python
from astroquery.gaia import Gaia
from astropy.coordinates import SkyCoord
import astropy.units as u

# 2MASS J23335840+0050119 coordinates (J2000)
coord = SkyCoord("23h33m58.40s", "+00d50m11.9s", frame="icrs")

# Search radius: 20 arcsec (adjust as needed)
radius = 20 * u.arcsec

# ADQL query: Gaia DR3 sources within radius, G <= 19
query = f"""
SELECT
    source_id, ra, dec, phot_g_mean_mag,
    phot_bp_mean_mag, phot_rp_mean_mag
FROM gaiadr3.gaia_source
WHERE 1 = CONTAINS(
    POINT('ICRS', ra, dec),
    CIRCLE('ICRS', {coord.ra.deg}, {coord.dec.deg}, {radius.to(u.deg).value})
)
AND phot_g_mean_mag <= 19
"""

job = Gaia.launch_job_async(query)
result = job.get_results()
print(result)
print(f"Number of stars with G <= 19: {len(result)}")
```
